**Question:** We continue to consider the use of a logistic regression model to predict the probability of `default` using `income` and `balance` on the `Default` data set. We will now compute estimates for the standard errors of the `income` and `balance` logistic regression coefficients in two different ways: (1) using the bootstrap, and (2) using the standard formula for computing the standard errors in the `sm.GLM()` function.

## Imports and Data

```python
import numpy as np                                    # numerical operations, arrays, random sampling
import statsmodels.api as sm                          # statistical models, including GLM for logistic regression
from ISLP import load_data                            # helper to load the book's built-in datasets
from ISLP.models import ModelSpec as MS, summarize     # ModelSpec builds design matrices; summarize() prints clean coefficient tables

Default = load_data('Default')                         # load the Default dataset (10,000 rows: default, student, balance, income)
Default.head()                                         # display the first 5 rows to confirm it loaded correctly
```

## (a) Standard errors using `summarize()` and `sm.GLM()`

We fit a multiple logistic regression model of `default` on `income` and `balance`, and read off the standard errors reported by `statsmodels`.

```python
X = MS(['income', 'balance']).fit_transform(Default)   # build the design matrix (intercept + income + balance columns)
y = (Default['default'] == 'Yes').astype(float)        # convert the Yes/No response into 1.0/0.0 for logistic regression

glm_model = sm.GLM(y, X, family=sm.families.Binomial()) # specify a logistic regression as a GLM with a Binomial family
glm_results = glm_model.fit()                           # fit the model via maximum likelihood
summarize(glm_results)                                  # display coefficients, standard errors, z-stats, and p-values
```

The `std err` column above gives the formula-based standard errors for `income` and `balance`. These come from the estimated covariance matrix of the maximum likelihood estimator, which relies on standard asymptotic theory (i.e. large-sample approximations).

## (b) `boot_fn()`

A function that takes the `Default` data set and a vector of indices, refits the logistic regression model on `data.iloc[idx]`, and returns the `income` and `balance` coefficient estimates.

```python
def boot_fn(data, idx):
    """Fit logistic regression of default ~ income + balance on data.iloc[idx],
    returning the income and balance coefficient estimates."""
    D = data.iloc[idx]                                  # select the (possibly repeated) rows given by idx -- this is the resampled dataset
    X_ = MS(['income', 'balance']).fit_transform(D)     # build the design matrix for this resampled dataset
    y_ = (D['default'] == 'Yes').astype(float)          # convert this resample's response to 1.0/0.0
    results_ = sm.GLM(y_, X_, family=sm.families.Binomial()).fit()   # refit logistic regression on the resampled data
    return results_.params[['income', 'balance']]       # return only the two coefficients we care about (drop the intercept)

# Sanity check: running boot_fn on the full (unresampled) index set
# should reproduce the coefficients from part (a)
full_idx = np.arange(Default.shape[0])                  # indices 0, 1, 2, ..., n-1 -- i.e. every row exactly once, no resampling
boot_fn(Default, full_idx)                               # run boot_fn on the full dataset as a sanity check
```

## (c) Estimating standard errors via the bootstrap

We repeatedly generate bootstrap samples (resampling `n` observations **with replacement**), apply `boot_fn()` to each, and compute the standard deviation of the resulting coefficient estimates across `B = 1000` replications.

```python
rng = np.random.default_rng(0)   # create a random number generator with a fixed seed, so results are reproducible
n = Default.shape[0]              # n = number of observations in the dataset (10,000)
B = 1000                          # number of bootstrap replications to run

boot_coefs = np.zeros((B, 2))     # pre-allocate an array to store the income/balance coefficient from each bootstrap replication
for b in range(B):                                       # repeat the resample-and-refit process B times
    idx = rng.choice(n, size=n, replace=True)             # draw n indices WITH replacement -- this is the bootstrap resample
    coefs = boot_fn(Default, idx)                         # refit the model on this resampled dataset and get its coefficients
    boot_coefs[b, 0] = coefs['income']                    # store this replication's income coefficient in column 0
    boot_coefs[b, 1] = coefs['balance']                   # store this replication's balance coefficient in column 1

boot_se_income  = boot_coefs[:, 0].std(ddof=1)   # standard deviation across all B income estimates = bootstrap SE for income
boot_se_balance = boot_coefs[:, 1].std(ddof=1)   # standard deviation across all B balance estimates = bootstrap SE for balance

print(f"Bootstrap SE (income) : {boot_se_income:.8f}")    # display the bootstrap standard error for income
print(f"Bootstrap SE (balance): {boot_se_balance:.8f}")   # display the bootstrap standard error for balance
```

## (d) Comparing the two sets of standard errors

```python
import pandas as pd                                    # for building a clean side-by-side comparison table

comparison = pd.DataFrame({
    'GLM formula SE': [glm_results.bse['income'], glm_results.bse['balance']],   # formula-based SEs from part (a)
    'Bootstrap SE':   [boot_se_income, boot_se_balance]                          # bootstrap-based SEs from part (c)
}, index=['income', 'balance'])                          # label the rows by predictor name

comparison                                                # display the comparison table
```

**Comment:**

The bootstrap standard errors and the `sm.GLM()` formula-based standard errors are very close to one another (within roughly 1-5%) for both `income` and `balance`.

This agreement is expected. The formula used by `sm.GLM()` relies on asymptotic theory for maximum likelihood estimation — it assumes the model is correctly specified and that the sample size is large enough for the sampling distribution of the coefficients to be approximately normal. With `n = 10,000` observations, these assumptions hold reasonably well, so the two approaches converge to similar answers.

The key conceptual difference is that the bootstrap standard errors are obtained **without relying on these distributional assumptions** — they come purely from resampling the observed data and directly measuring how much the coefficient estimates vary across resamples. This makes the bootstrap especially valuable in settings where the standard formula's assumptions are questionable (e.g., small sample sizes, or models where no formula for the standard error exists at all). The fact that the two methods agree closely here serves as a useful validation of both approaches for this particular problem.
